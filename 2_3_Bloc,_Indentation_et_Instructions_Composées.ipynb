```python
def sum_1_to_n(n):
    """Somme de 1 jusqu'à n"""
    sum = 0
    i = 0
    while (i <= n):
        sum += i
        i += 1
    return sum

```

```python
print(sum_1_to_n(1))  # Appel de la fonction
```

```python
print(sum_1_to_n(100))
```
